```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime
from IPython.display import display, Markdown

todays_date = datetime.now().strftime("%Y-%m-%d")
student_id = "vingut8011"

display(Markdown("## Loan Scenario Analysis"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

principal = 300000

scenarios = [
    {"Scenario": "3% for 15 Years", "Rate": 0.03, "Years": 15},
    {"Scenario": "5% for 30 Years", "Rate": 0.05, "Years": 30},
    {"Scenario": "8% for 40 Years", "Rate": 0.08, "Years": 40}
]

def calculate_payment(P, annual_rate, years):
    r = annual_rate / 12
    n = years * 12

    M = P * r * (1 + r)**n / ((1 + r)**n - 1)
    return M

results = []

for scenario in scenarios:
    monthly_payment = calculate_payment(
        principal,
        scenario["Rate"],
        scenario["Years"]
    )

    total_repayment = monthly_payment * scenario["Years"] * 12

    results.append([
        scenario["Scenario"],
        round(monthly_payment, 2),
        round(total_repayment, 2)
    ])

results_df = pd.DataFrame(
    results,
    columns=["Scenario", "Monthly Payment", "Total Repayment"]
)

display(results_df)
```

```python
display(Markdown("## Repayment Comparison"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

plt.figure(figsize=(8,5))

plt.bar(
    results_df["Scenario"],
    results_df["Total Repayment"]
)

plt.title("Total Loan Repayment by Scenario")
plt.xlabel("Scenario")
plt.ylabel("Total Repayment ($)")
plt.xticks(rotation=15)

plt.show()
```

```python
display(Markdown("## Sensitivity Analysis"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

modified_rate = 0.06

monthly_payment_modified = calculate_payment(
    principal,
    modified_rate,
    30
)

total_modified = monthly_payment_modified * 30 * 12

comparison_df = pd.DataFrame({
    "Scenario": ["Original 5% 30 Years", "Modified 6% 30 Years"],
    "Total Repayment": [
        results_df.iloc[1]["Total Repayment"],
        total_modified
    ]
})

display(comparison_df)

plt.figure(figsize=(8,5))

plt.bar(
    comparison_df["Scenario"],
    comparison_df["Total Repayment"]
)

plt.title("Sensitivity Analysis")
plt.ylabel("Total Repayment ($)")

plt.show()
```

## Analyze and Interpret

### 1. What differences were observed between the scenarios?

The scenarios produced different monthly payment and total repayment amounts. The 3% interest rate with a 15-year term resulted in the lowest total repayment amount. The 5% interest rate with a 30-year term increased the overall repayment amount, while the 8% interest rate with a 40-year term resulted in the highest total repayment because of both the higher interest rate and longer loan term.

### 2. How did changes in key variables affect the outcomes in the sensitivity analysis?

The sensitivity analysis showed that increasing the interest rate increased the total repayment amount. Even a small increase in the interest rate significantly affected the overall cost of the loan over time. This demonstrates that interest rates are a critical factor in determining the affordability of a loan.
